```
# Encode Categorical Variables

X_encoded = pd.get_dummies(
    X,
    columns=['Geography', 'Gender'],
    drop_first=True)

print("Columns after encoding:")
print(X_encoded.columns.tolist())

print("\nShape of encoded features:")
print(X_encoded.shape)

print("\nFirst 5 rows:")
print(X_encoded.head())
```

# **Threshold Analysis**

```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_score, recall_score, f1_score

# Prepare modeling data
model_df = df.drop(
    columns=['Year', 'CustomerId', 'Surname'],
    errors='ignore'
).copy()

model_df = model_df.drop(
    columns=['Age_Group', 'Balance_Group'],
    errors='ignore'
)

# Separate features and target
X = model_df.drop(columns=['Exited'])
y = model_df['Exited']

# One-hot encoding
X_encoded = pd.get_dummies(
    X,
    columns=['Geography', 'Gender'],
    drop_first=True
)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Train Random Forest
rf_final = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)

rf_final.fit(X_train, y_train)

# Churn probability
churn_prbability = rf_final.predict_proba(X_test)[:, 1]

# Test threshold
thresholds = [0.30, 0.40, 0.50, 0.60]

results = []

for threshold in thresholds:
    prediction = (churn_prbability >= threshold).astype(int)

    results.append({
        'Threshold': threshold,
        'Precision': precision_score(
            y_test, prediction, zero_division=0),
        'Recall': recall_score(
            y_test, prediction, zero_division=0),
        'F1-Score': f1_score(
            y_test, prediction, zero_division=0)
    })

    threshold_df = pd.DataFrame(results)

print("Random Forest Threshold Analysis:")
display(threshold_df.round(4))
```

```
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Apply selected threshold
final_prediction = (
    churn_prbability >= 0.30
).astype(int)

# Create confusion matrix
cm = confusion_matrix(y_test, final_prediction)

print("Confusion Matrix:")
print(cm)

# Display confusion matrix
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Retained', 'Churned']
)

disp.plot()

plt.title('Random Forest Confusion Matrix - Threshold 0.30')
plt.tight_layout()
plt.show()
```

```
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

final_accuracy = accuracy_score(
    y_test,
    final_prediction
)

final_precision = precision_score(
    y_test,
    final_prediction,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    final_prediction,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    final_prediction,
    zero_division=0
)

print("Final Random Forest Model Performance:")
print("--------------------------------------")
print("Threshold :", 0.30)
print("Accuracy :", round(final_accuracy, 4))
print("Precision :", round(final_precision, 4))
print("Recall :", round(final_recall, 4))
print("F1-Score:", round(final_f1, 4))
```

# **Final Retention Strategy**

## Final Customer Retention Strategy

Based on the exploratory analysis and Random Forest Churn prediction model, the following retention strategies are recommended:

### 1. Priorotize High-Risk Customers
Customers classified as High-Risk should receive proactive retention interventions before churn occurs.

### 2. Re-engage Inactive Customers
Most high-risk customers are inactive members (125 out of 160).
Banks should use personlized communication, service reminders, and relevant product offers to increase customer activity.

### 3. Focus on the 46-55 Age Segment
The 46-55 age group represents the largest share of high-risk customers. Retention programs should provide relevant fianancial products and personlized relationship management for this segment.

### 4. Strenthen Engagement in Germany
Germany has the largest number of high-risk customers (86 out of 160). Customer engagement and service initiative should receive particular attention in this market.

### 5. Improve Product Utilization
High-risk customers are concentrated among customers with one product and also include a notable group three or four products.
Banks should focus on understnading product usage and offering relevant products rather than simply increasing the number of products.

### 6. Use Prediction Churn Scoring
The Random Forest model can be used to assign customers to low, Medium, and High Risk categories. The selected probability threshold of 0.30 prioritizes identifying potential churners.

### 7. Continuous Monitoring
Customer churn probabitlities should be periodically updated so that retention teams can identify changes in customer risk and take timely action.
